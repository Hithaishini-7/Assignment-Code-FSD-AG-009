Question 1 : In simple words, what are React Hooks? Create a small component that
shows a button. When the button is clicked, toggle the text between “On” and “Off”.

Answer-React Hooks are special functions in React that let us use features like state in functional components.
In simple words, Hooks help a component remember information and update the screen when something changes.
For example:
- useState helps us store and change data.
- useEffect helps us perform tasks when a component loads or changes.

```
import React, { useState } from 'react';

function ToggleButton() {
  const [status, setStatus] = useState("Off");

  function toggleStatus() {
    setStatus(status === "Off" ? "On" : "Off");
  }

  return (
    <div>
      <h2>{status}</h2>

      <button onClick={toggleStatus}>
        Click Me
      </button>
    </div>
  );
}

export default ToggleButton;
```

changed” using useEffect.

Answer-

```
import React, { useState, useEffect } from 'react';

function Counter() {
  const [count, setCount] = useState(0);

  useEffect(() => {
    console.log("Count changed");
  }, [count]);

  return (
    <div>
      <h2>Count: {count}</h2>

      <button onClick={() => setCount(count + 1)}>
        Increase
      </button>
    </div>
  );
}

export default Counter;
```

### Simple Explanation

1. `useState(0)` – Sets the initial count to `0`.
2. `useEffect()` – Runs a function when the component loads or when the specified value changes.
3. `[count]` – Tells React to run the effect when `count` changes.
4. `console.log("Count changed")` – Prints `"Count changed"` in the browser console.
5. `onClick` – Increases the count by 1 whenever the button is clicked.

### Output

## Count: 0

Increase

Browser console: "Count changed"

Note: The `useEffect` runs once when the component first loads and again whenever `count` changes. In development mode, React Strict Mode may cause the effect to run an extra time on the initial load.



Question 3 : Write a component that fetches users from
https://jsonplaceholder.typicode.com/users only once on mount and
shows how many users were loaded.

Answer-## Question 3: Fetch Users using `useEffect`

### Code

```
import React, { useState, useEffect } from 'react';

function Users() {
  const [users, setUsers] = useState([]);

  useEffect(() => {
    fetch('https://jsonplaceholder.typicode.com/users')
      .then(response => response.json())
      .then(data => setUsers(data));
  }, []);

  return (
    <div>
      <h2>Number of Users: {users.length}</h2>
    </div>
  );
}

export default Users;
```

### Simple Explanation

1. `useState([])` – Creates an empty array to store users.
2. `useEffect()` – Runs code when the component loads.
3. `fetch()` – Gets user data from the given URL.
4. `response.json()` – Converts the response into JSON data.
5. `setUsers(data)` – Stores the fetched users in the state.
6. `users.length` – Displays the total number of users.
7. – Ensures the effect runs only when the component first mounts (loads).

### Output

Initially, it shows:

`Number of Users: 0`

After the data loads, it shows:

`Number of Users: 10`

Remember: An empty dependency array `[]` makes `useEffect` run only on the initial mount (apart from React Strict Mode's extra development checks).


Question 4 : Create a component that starts a timer with setInterval to increase a
“seconds” counter every second. Stop the timer when the component unmounts.

Answer-## Question 4: Timer using `setInterval` and `useEffect`

### Code

```
import React, { useState, useEffect } from 'react';

function Timer() {
  const [seconds, setSeconds] = useState(0);

  useEffect(() => {
    const timer = setInterval(() => {
      setSeconds(prevSeconds => prevSeconds + 1);
    }, 1000);

    return () => {
      clearInterval(timer);
    };
  }, []);

  return (
    <div>
      <h2>Seconds: {seconds}</h2>
    </div>
  );
}

export default Timer;
```

### Simple Explanation

1. `useState(0)` – Starts the seconds counter at `0`.
2. `setInterval()` – Runs a function every 1 second (1000 milliseconds).
3. `setSeconds(prevSeconds => prevSeconds + 1)` – Increases the counter by 1 every second.
4. `clearInterval(timer)` – Stops the timer when the component unmounts (is removed from the screen).
5. `useEffect(..., [])` – Starts the timer when the component loads.

### Output

The counter increases automatically:

```
Seconds: 0
Seconds: 1
Seconds: 2
Seconds: 3
Seconds: 4
...
```

Remember: `setInterval()` starts the timer, and `clearInterval()` stops it. The cleanup function returned by `useEffect` prevents the timer from running after the component is removed.

Question 5 : Build a small form with a text input. If the input is empty, show “Please
type your name”. If it has text, show “Hello, <name>”.

Answer-
## Question 5: Form with Text Input in React

### Code

```
import React, { useState } from 'react';

function NameForm() {
  const [name, setName] = useState("");

  return (
    <div>
      <h2>Enter Your Name</h2>

      <input
        type="text"
        value={name}
        onChange={(e) => setName(e.target.value)}
        placeholder="Type your name"
      />

      <h3>
        {name.trim() === ""
          ? "Please type your name"
          : `Hello, ${name}`}
      </h3>
    </div>
  );
}

export default NameForm;
```

### Simple Explanation

1. `useState("")` – Stores the name. Initially, it is empty.
2. `value={name}` – Displays the current name in the input box.
3. `onChange` – Updates the name whenever the user types.
4. `name.trim() === ""` – Checks whether the input is empty or contains only spaces.
5. `? :` – This is a ternary operator that chooses which message to display.
   - If the input is empty, it displays "Please type your name".
   - If the input contains a name, it displays "Hello, Rahul" (or whatever name you type).

### Output

### Enter Your Name

Please type your name

Remember: `useState` stores the input value, and `onChange` updates it whenever the user types.

Question 6 : You have a theme value (“light” or “dark”) that is needed by a deep child
button. Show how to pass it without sending theme through every parent prop

Answer-
## Question 6: Using Context API in React

To pass a theme (`"light"` or `"dark"`) to a deeply nested child without passing it through every parent component, we use Context API.

React provides `createContext()` and `useContext()` to share data between components.

### Code

```
import React, { createContext, useContext } from 'react';

// Create Context
const ThemeContext = createContext();

// Child Button
function ChildButton() {
  const theme = useContext(ThemeContext);

  return (
    <button className={theme}>
      {theme} Theme Button
    </button>
  );
}

// Parent Component
function Parent() {
  return (
    <div>
      <ChildButton />
    </div>
  );
}

// Main Component
function App() {
  return (
    <ThemeContext.Provider value="dark">
      <Parent />
    </ThemeContext.Provider>
  );
}

export default App;
```

### Simple Explanation

1. `createContext()` – Creates a context to share the theme.
2. `ThemeContext.Provider` – Makes the theme value `"dark"` available to child components inside it.
3. `useContext(ThemeContext)` – Allows `ChildButton` to access the theme directly.
4. `Parent` – Does not need to receive or pass the theme as a prop.

### Output

The button displays:

`dark Theme Button`

Remember: Context API helps share data with deeply nested components without passing props through every parent. This problem is commonly called prop drilling.

Question 7 : Make a counter with two buttons: “+1” and “-1”. Use useReducer instead
of useState.
Answer-
## Question 7: Counter using `useReducer`

### Code

```
import React, { useReducer } from 'react';

function reducer(count, action) {
  if (action === "INCREASE") {
    return count + 1;
  }

  if (action === "DECREASE") {
    return count - 1;
  }

  return count;
}

function Counter() {
  const [count, dispatch] = useReducer(reducer, 0);

  return (
    <div>
      <h2>Count: {count}</h2>

      <button onClick={() => dispatch("INCREASE")}>
        +1
      </button>

      <button onClick={() => dispatch("DECREASE")}>
        -1
      </button>
    </div>
  );
}

export default Counter;
```

### Simple Explanation

1. `useReducer` – Manages the counter value.
2. `reducer` – A function that decides how to change the count.
3. `count` – Stores the current counter value, starting at `0`.
4. `dispatch("INCREASE")` – Sends an action to increase the count by 1.
5. `dispatch("DECREASE")` – Sends an action to decrease the count by 1.

### Output

# Count: 0

+1-1

Remember: `useReducer` is useful when state changes depend on different actions. Here, the actions are `"INCREASE"` and `"DECREASE"`.

Question 8 : Given an array nums = [1,2,3,4] and a filter value min, show how to
compute a filtered list only when nums or min change. Also pass a stable click handler
to a child list item.
Answer-
## Question 8: Using `useMemo` and `useCallback`

### Code

```
import React, { useState, useMemo, useCallback } from 'react';

function ListItem({ number, onClick }) {
  return (
    <li onClick={() => onClick(number)}>
      {number}
    </li>
  );
}

function App() {
  const nums = [1, 2, 3, 4];
  const [min, setMin] = useState(2);

  // Filter the list only when nums or min changes
  const filteredNums = useMemo(() => {
    return nums.filter(num => num >= min);
  }, [nums, min]);

  // Create a stable click handler
  const handleClick = useCallback((num) => {
    console.log("Clicked:", num);
  }, []);

  return (
    <div>
      <h2>Filtered Numbers</h2>

      <button onClick={() => setMin(min + 1)}>
        Increase Minimum
      </button>

      <ul>
        {filteredNums.map(num => (
          <ListItem
            key={num}
            number={num}
            onClick={handleClick}
          />
        ))}
      </ul>
    </div>
  );
}

export default App;
```

### Simple Explanation

1. `useMemo` – Saves the filtered list and recalculates it when `nums` or `min` changes.
2. `filter()` – Keeps only numbers greater than or equal to `min`.
3. `useCallback` – Keeps the same function reference between renders as long as its dependencies do not change.
4. `handleClick` – Runs when a list item is clicked and prints the clicked number in the console.
5. `key={num}` – Helps React identify each list item.

### Output

Initially, `min = 2`, so the list displays:

```
Filtered Numbers

2
3
4
```

After clicking Increase Minimum, `min` becomes `3`, and the list displays:

```
3
4
```

### Remember

- `useMemo` memoizes a value.
- `useCallback` memoizes a function.

Important note: In this example, `nums` is created inside the component, so it gets a new array reference on every render. To truly avoid unnecessary recalculation, move it outside the component:

```
const nums = [1, 2, 3, 4];
```

Question 9 : Create a controlled input for email. As the user types, show “Email:
<value>”. Why do we keep the input’s value in state?
Answer-
## Question 9: Controlled Input for Email

### Code

```
import React, { useState } from 'react';

function EmailInput() {
  const [email, setEmail] = useState("");

  return (
    <div>
      <input
        type="email"
        value={email}
        onChange={(e) => setEmail(e.target.value)}
        placeholder="Enter your email"
      />

      <p>Email: {email}</p>
    </div>
  );
}

export default EmailInput;
```

### Simple Explanation

1. `useState("")` – Stores the email. Initially, it is empty.
2. `value={email}` – Displays the email stored in state inside the input.
3. `onChange` – Runs whenever the user types something.
4. `setEmail(e.target.value)` – Updates the email in state.
5. `Email: {email}` – Displays the email as the user types.

### Output

If the user types `abc@gmail.com`, the screen displays:

```
[ abc@gmail.com ]

Email: abc@gmail.com
```

### Why do we keep the input's value in state?

We keep the input value in state because:

- To store the data: React remembers what the user types.
- To update the screen: The displayed email changes automatically.
- To use the data: We can validate the email or submit it through a form.


Question 10 : Make a button that focuses an input using a ref. Return adjacent
elements without extra wrapper nodes. In one sentence, say what React’s update
process does\
Answer-
## Question 10: Focus an Input Using `useRef`

### Code

```
import React, { useRef } from 'react';

function FocusInput() {
  const inputRef = useRef(null);

  function handleFocus() {
    inputRef.current.focus();
  }

  return (
    <>
      <input ref={inputRef} type="text" placeholder="Enter text" />
      <button onClick={handleFocus}>Focus Input</button>
    </>
  );
}

export default FocusInput;
```

### Simple Explanation

1. `useRef(null)` – Creates a reference to access the input element.
2. `ref={inputRef}` – Connects the reference to the input.
3. `inputRef.current.focus()` – Places the cursor inside the input when the button is clicked.
4. `<> </>` – A React Fragment that lets us return adjacent elements without adding an extra wrapper node to the DOM.

### What does React’s update process do?

React’s update process checks what has changed in the component and updates the necessary parts of the DOM to reflect those changes.
